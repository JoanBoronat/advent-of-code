### [Day 20: Jurassic Jigsaw](https://adventofcode.com/2020/day/20)

The high-speed train leaves the forest and quickly carries you south. You can even see a desert in the distance! Since you have some spare time, you might as well see if there was anything interesting in the image the Mythical Information Bureau satellite captured.

 After decoding the satellite messages, you discover that the data actually contains many small images created by the satellite's **camera array**. The camera array consists of many cameras; rather than produce a single square image, they produce many smaller square image **tiles** that need to be **reassembled back into a single image**.

 Each camera in the camera array returns a single monochrome **image tile** with a random unique **ID number**. The tiles (your puzzle input) arrived in a random order.

 Worse yet, the camera array appears to be malfunctioning: each image tile has been **rotated and flipped to a random orientation**. Your first task is to reassemble the original image by orienting the tiles so they fit together.

 To show how the tiles should be reassembled, each tile's image data includes a border that should line up exactly with its adjacent tiles. All tiles have this border, and the border lines up exactly when the tiles are both oriented correctly. Tiles at the edge of the image also have this border, but the outermost edges won't line up with any other tiles.

````python
import re

# Read the input data
with open('data/advent_20.txt', 'r') as f:
    data = f.read().strip()

# Parse tiles
tiles = {}
tile_blocks = data.split('\n\n')

for block in tile_blocks:
    lines = block.strip().split('\n')
    # Extract tile ID from first line
    tile_id = int(re.search(r'Tile (\d+):', lines[0]).group(1))
    # Extract the grid (10x10)
    grid = [list(line) for line in lines[1:]]
    tiles[tile_id] = grid

# Save a copy of original tiles for later use
original_tiles = {tid: grid for tid, grid in tiles.items()}
````

````python
def get_borders(tile_grid):
    size = len(tile_grid)
    top = ''.join(tile_grid[0])
    right = ''.join(tile_grid[i][size-1] for i in range(size))
    bottom = ''.join(tile_grid[size-1])
    left = ''.join(tile_grid[i][0] for i in range(size))    
    return [top, right, bottom, left]
````

````python
def index_to_direction(index):
    return [(0, -1), (1, 0), (0, 1), (-1, 0)][index]
````

````python
def flip_tile_horizontal(tile_grid):
    """Flip a tile horizontally (reverse each row)"""
    return [row[::-1] for row in tile_grid]

def rotate90(tile_grid):
    """Rotate a tile 90 degrees clockwise"""
    # Handle both list of lists and list of strings
    if isinstance(tile_grid[0], str):
        # If it's a list of strings, convert to list of lists first
        tile_grid = [list(row) for row in tile_grid]
    
    # Transpose and reverse each row
    size = len(tile_grid)
    rotated = [[tile_grid[size - 1 - j][i] for j in range(size)] for i in range(size)]
    
    return rotated

def orientations(tile):
    """Generate all 4 rotations of a tile"""
    yield tile
    for _ in range(3):
        tile = rotate90(tile)
        yield tile

def arrangements(tile):
    """Generate all 8 possible arrangements (4 rotations × 2 flips)"""
    yield from orientations(tile)
    yield from orientations(tile[::-1])  # Flip vertically (reverse rows)

def get_opposite_index(index):
    """Get the opposite side index (0=top↔2=bottom, 1=right↔3=left)"""
    return {0: 2, 1: 3, 2: 0, 3: 1}[index]

def edge(tile_grid, side):
    """Get the edge string for a given side (n=north/top, s=south/bottom, e=east/right, w=west/left)"""
    size = len(tile_grid)
    if side == 'n' or side == 0:  # top
        return ''.join(tile_grid[0])
    elif side == 'e' or side == 1:  # right
        return ''.join(tile_grid[i][size-1] for i in range(size))
    elif side == 's' or side == 2:  # bottom
        return ''.join(tile_grid[size-1])
    elif side == 'w' or side == 3:  # left
        return ''.join(tile_grid[i][0] for i in range(size))

def matching_tile(tile, tiles_dict, side_a, side_b):
    """
    Find a tile that matches the given tile on the specified sides.
    Returns the oriented tile and removes it from tiles_dict.
    """
    prev_side = edge(tile, side_a)
    
    # Iterate over all possible tiles
    for other_id, other in list(tiles_dict.items()):
        if tile is other:
            continue
        
        # Try all 8 arrangements of the candidate tile
        for arranged in arrangements(other):
            # Check if the sides match
            if prev_side == edge(arranged, side_b):
                tiles_dict.pop(other_id)
                return arranged
    
    return None

def check_neighbors(current_tile_id, tile_id):
    """
    Check if tile_id can be placed next to current_tile_id.
    Returns the border index (0-3) if a match is found, None otherwise.
    This function orients the tile correctly if a match is found.
    """
    current_tile = tiles[current_tile_id]
    candidate_tile = tiles[tile_id]
    
    # Check all 4 directions (borders) of the current tile
    for border_idx in range(4):
        # Get the edge we want to match
        current_edge = edge(current_tile, border_idx)
        
        # Try all 8 arrangements of the candidate tile
        for arranged in arrangements(candidate_tile):
            # Check if the opposite side matches
            opposite_side = get_opposite_index(border_idx)
            candidate_edge = edge(arranged, opposite_side)
            
            if current_edge == candidate_edge:
                # Found a match! Orient the tile correctly
                tiles[tile_id] = arranged
                return border_idx
    
    return None

````

````python
position = (0, 0)
first_tile_id = list(tiles.keys())[0]
positions = {position: first_tile_id}
tiles_placed = [first_tile_id]
i = 0

while len(tiles) != len(tiles_placed):
    if i >= len(tiles_placed):
        raise Exception("No more tiles to place")
        
    current_tile_id = tiles_placed[i]
    # Find the position of the current tile
    current_position = None
    for pos, tid in positions.items():
        if tid == current_tile_id:
            current_position = pos
            break
    
    # Skip tiles that don't have positions yet (they'll be processed later)
    if current_position is None:
        i += 1
        continue
    
    for tile_id, tile_grid in tiles.items():
      if tile_id in tiles_placed:
        continue
      # Reset tile to original before checking (it might have been modified in previous checks)
      tiles[tile_id] = [row[:] for row in original_tiles[tile_id]]
      matched_border = check_neighbors(current_tile_id, tile_id)

      if matched_border is not None:
        vector = index_to_direction(matched_border)
        new_position = (current_position[0] + vector[0], current_position[1] + vector[1])
        
        # Check if position is already occupied
        if new_position in positions:
          continue
        
        positions[new_position] = tile_id
        tiles_placed.append(tile_id)
        print(f"{current_tile_id} -> {tile_id} at {new_position}")
    
    i += 1

````

````python
# Find corner tiles using min/max coordinates
x_coords = [pos[0] for pos in positions.keys()]
y_coords = [pos[1] for pos in positions.keys()]

min_x, max_x = min(x_coords), max(x_coords)
min_y, max_y = min(y_coords), max(y_coords)

edges = positions[min_x, min_y], positions[max_x, min_y], positions[min_x, max_y], positions[max_x, max_y]

result = edges[0] * edges[1] * edges[2] * edges[3]
print(result)
````
